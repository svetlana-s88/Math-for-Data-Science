```python
import matplotlib.pyplot as plt
import numpy as np


def sigmoid(x):
    return 1 / (1 + np.exp(-x))


# Генерируем 200 точек от -5 до 5
x = np.linspace(-5, 5, 200) 
y = sigmoid(x)

# Настройка графика
plt.figure(figsize=(8, 5)) # задаем размеры графика
plt.plot(x, y, label=r"$\sigma(x) = \frac{1}{1 + e^{-x}}$", color="blue", lw=1)

# Добавляем сетку и оси для наглядности
plt.axhline(0, color="black", linewidth=0.5, linestyle="--")
plt.axhline(0.5, color="gray", linewidth=0.5, linestyle="--")
plt.axhline(1, color="black", linewidth=0.5, linestyle="--")
plt.axvline(0, color="black", linewidth=0.5)

# Заголовки и подписи
plt.title("График логистической функции (Сигмоида)")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True, alpha=0.3)
plt.legend(fontsize=12)

# Отображение
plt.show()
```
